```python

import pandas as pd

# Load the sample dataset
df = pd.read_csv("../data/flight_data_2024_sample.csv")

# Display the first 5 rows
df.head()
```

```python

# Check the number of rows and columns
print("Dataset Shape:", df.shape)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

# Display dataset information
df.info()
```

```python

# Check missing values in each column
missing_values = df.isnull().sum()

# Display columns with missing values
print(missing_values[missing_values > 0])
```

```python

# Check duplicate records
duplicates = df.duplicated().sum()

print("Total Duplicate Records:", duplicates)
```

```python

# Display statistical summary
df.describe()

```

```python

# Check flight delay distribution
valid_flights = df[df["dep_delay"].notna()].copy()

valid_flights["delayed"] = (
    valid_flights["dep_delay"] >= 15
).astype(int)

print("Flight Delay Distribution:")
print(valid_flights["delayed"].value_counts())

print("\nDelay Percentage:")
print(valid_flights["delayed"].value_counts(normalize=True) * 100)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Plot flight delay distribution
plt.figure(figsize=(7, 5))

sns.countplot(
    data=valid_flights,
    x="delayed"
)

plt.title("Flight Delay Distribution")
plt.xlabel("Flight Status (0 = Not Delayed, 1 = Delayed)")
plt.ylabel("Number of Flights")

plt.show()
```

```python

# Calculate flight delay percentage by month
monthly_delay = (
    valid_flights.groupby("month")["delayed"].mean() * 100
)

# Plot monthly delay percentage
plt.figure(figsize=(10, 5))

sns.barplot(
    x=monthly_delay.index,
    y=monthly_delay.values
)

plt.title("Flight Delay Percentage by Month")
plt.xlabel("Month")
plt.ylabel("Delayed Flights (%)")
plt.ylim(0, 100)

plt.show()
```

```python

# Calculate delay percentage by airline
airline_delay = (
    valid_flights.groupby("op_unique_carrier")["delayed"]
    .agg(["mean", "count"])
)

airline_delay["delay_percentage"] = airline_delay["mean"] * 100

# Display airline delay percentages
print(
    airline_delay[["count", "delay_percentage"]]
    .sort_values("delay_percentage", ascending=False)
)
```

```python

# Sort airlines by delay percentage
airline_delay_sorted = airline_delay.sort_values(
    "delay_percentage",
    ascending=False
)

# Create bar chart
plt.figure(figsize=(12, 6))

sns.barplot(
    x=airline_delay_sorted.index,
    y=airline_delay_sorted["delay_percentage"]
)

plt.title("Flight Delay Percentage by Airline")
plt.xlabel("Airline Code")
plt.ylabel("Delayed Flights (%)")
plt.xticks(rotation=45)
plt.ylim(0, 100)

plt.show()
```

```python

# Extract scheduled departure hour
valid_flights["departure_hour"] = (
    valid_flights["crs_dep_time"] // 100
).astype(int)

# Calculate delay percentage by hour
hourly_delay = (
    valid_flights.groupby("departure_hour")["delayed"]
    .mean() * 100
)

# Plot hourly delay percentage
plt.figure(figsize=(12, 5))

sns.lineplot(
    x=hourly_delay.index,
    y=hourly_delay.values,
    marker="o"
)

plt.title("Flight Delay Percentage by Departure Hour")
plt.xlabel("Scheduled Departure Hour")
plt.ylabel("Delayed Flights (%)")
plt.xticks(range(0, 24))
plt.ylim(0, 100)

plt.show()
```

```python

# Count flights and calculate delay percentage by hour
hourly_summary = valid_flights.groupby(
    "departure_hour"
)["delayed"].agg(["count", "mean"])

hourly_summary["delay_percentage"] = (
    hourly_summary["mean"] * 100
)

print(hourly_summary[["count", "delay_percentage"]])
```

```python

# Keep hours with at least 30 flights for visualization
filtered_hours = hourly_summary[
    hourly_summary["count"] >= 30
]

# Create improved chart
plt.figure(figsize=(12, 5))

sns.lineplot(
    x=filtered_hours.index,
    y=filtered_hours["delay_percentage"],
    marker="o"
)

plt.title("Flight Delay Percentage by Departure Hour")
plt.xlabel("Scheduled Departure Hour")
plt.ylabel("Delayed Flights (%)")
plt.xticks(range(5, 24))
plt.ylim(0, 50)
plt.grid(True)

plt.show()
```
